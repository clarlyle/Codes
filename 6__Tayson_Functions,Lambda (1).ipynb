## What is a Function

### Notes

* A **function** is a block of code that only runs when it's called.
* You can pass data (called **parameters**) into a function.
* The function can return data as a result.

## Advanced Concepts 

### Arbitrary Arguments (*Args)

#### Notes

* If you're not sure how many arguments will be passed into your function, add a `*` before the parameter name in the function definition.
* The function will now receive a tuple of arguments.
* Arbitrary Arguments are often called `*args`.




````python
def numbers(*args):
    print(args)

numbers(10, 20, 30)
````

#### Activity

- Define a function that accepts a variable number of arguments using *args
- Use for loop to print skills with their corresponding average salaries using positional arguments.
- Each argument is expected to be a tuple containing a job title or skill and its average salary.

        - :param args: A sequence of tuples, where each tuple contains a job title or skill and its average salary.

````python
def print_salaries(*args):
    """
    :param args: A sequence of tuples, where each tuple contains a job title or skill and its average salary.
    """
    for skill, salary in args:
        print(f"- {skill}: ${salary:,}")
````

- Call the function with tuples of job titles and their average salaries 
- Syntax is function_name(*args) where args is a tuple of arguments create 3 tuples of job titles and their average salaries and pass them to the function as arguments


````python
job1 = ("Data Analyst", 60000)
job2 = ("Business Analyst", 65000)
job3 = ("Data Scientist", 95000)

print_salaries(job1, job2, job3)
````

### Arbitrary Keyword Arguments (**Kwargs)

#### Notes

* If you don't know how many keyword arguments will be passed into the function, add two asterisk `**` before the parameter name in the function definition.
* This will have the function receive a dictionary of arguments.

````python
def show_info(**kwargs):
    print(kwargs)

show_info(name="Anna", age=30, subject="IT")
````

#### Activity
- Define a function that accepts a variable number of keyword arguments using **kwargs
- Use the loop to iterate through the keyword arguments and print each skill and its corresponding average salary

````python
# Function
def print_salaries(**kwargs):
    for skill, salary in kwargs.items():
        print(f"- {skill}: ${salary:,}")

# Call the function with keyword arguments for skills and their average salaries syntax is skill=salary, skill=salary, skill=salary, etc.
print_salaries(Data_Analyst=60000, Business_Analyst=65000, Data_Scientist=95000)
````

##### The difference of *args and **kwargs is that `*args` allows you to pass a variable number of positional arguments to a function, while   `**kwargs` allows you to pass a variable number of keyword arguments to a function. In other words, *args is used for non-keyword arguments, while **kwargs is used for keyword arguments.

# Lambda

## Notes

* `lambda` functions are small, anonymous functions.
* They can have any number of arguments but only one expression.
* The expression is evaluated and returned.
* It has the syntax of: `lambda arguments: expression`.
* Why would you use this?
    * Concise, it lets you write functions in a single line of code.
    * Since its anonymous it doesn't require a separate name like regular functions are defined using `def` keyword (it's good for when you need a function temporarily).


## Importance

Offers a succinct way to define anonymous functions. Useful in pandas for applying quick transformations to data.

````python
# Define a lambda function to add two numbers
add = lambda x, y: x + y
````

````python
# Use the lambda function to add two numbers
result1 = add(3, 5)
result1
````

## Example #2

Filter job postings based on the following:
- Job skills contains "Python"
- The job is remote (i.e., `True`)

````python
jobs_data = [
    {'job_title': 'Data Scientist',  'job_skills': ['Python', 'Machine Learning'], 'remote': True},
    {'job_title': 'Data Analyst',  'job_skills': ['Excel', 'SQL'], 'remote': False},
    {'job_title': 'Machine Learning Engineer', 'job_skills': ['Python', 'TensorFlow', 'Keras'], 'remote': True},
    {'job_title': 'Software Developer', 'job_skills': ['Java', 'C++'], 'remote': True},
    {'job_title': 'Data Scientist', 'job_skills': ['R', 'Statistics'], 'remote': False}
]
````

For this we will be using the `filter()` function

````python
help(filter)
````

 For the `filter()` function it takes two arguments:
 - function (we will use a lambda function)
 - iterable (we will use the job_data for this)
 
 |  filter(function or None, iterable) --> filter object
 |  
 |  Return an iterator yielding those items of iterable for which function(item)
 |  is true. If function is None, return the items that are true.

````python
# first let's just test out and only filter jobs that are remote

list(filter(lambda job: job['remote'], jobs_data))
````

Now let's make this more readable and add in the filter for 'Python'.

````python
# Define a lambda function for filtering jobs that require Python and are remote
is_python_remote = lambda job: 'Python' in job['job_skills'] and job['remote']

# Apply the filter
filtered_jobs = list(filter(is_python_remote, jobs_data))

# Display the filtered job listings
for job in filtered_jobs:
    print(job)
````
